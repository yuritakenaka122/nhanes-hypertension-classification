```python
# Libraries for data manipulation and visualization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

# Display all columns when examining datasets
pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

```python
# Define the folder containing NHANES datasets
data_dir = Path("Data")

# Load NHANES 2017–2018 data
demographics = pd.read_sas(data_dir / "DEMO_J.XPT")
body_measures = pd.read_sas(data_dir / "BMX_J.XPT")
blood_pressure = pd.read_sas(data_dir / "BPX_J.XPT")
bp_questionnaire = pd.read_sas(data_dir / "BPQ_J.XPT")

# Check dimensions
print("Demographics:", demographics.shape)
print("Body measurements:", body_measures.shape)
print("Blood pressure:", blood_pressure.shape)
print("BP questionnaire:", bp_questionnaire.shape)
```

```python
datasets = {
    "Demographics": demographics,
    "Body Measurements": body_measures,
    "Blood Pressure": blood_pressure,
    "BP Questionnaire": bp_questionnaire
}

for name, df in datasets.items():
    print(f"\n--- {name} ---")
    print("Shape:", df.shape)
    print("Unique participants:", df["SEQN"].nunique())
    print("Duplicate SEQN values:", df["SEQN"].duplicated().sum())
    print("Columns:", df.columns.tolist())
```

```python
# Preview selected demographic variables
display(
    demographics[
        ["SEQN", "RIDAGEYR", "RIAGENDR", "WTMEC2YR"]
    ].head()
)

# Preview body measurements
display(
    body_measures[
        ["SEQN", "BMXBMI", "BMXWAIST"]
    ].head()
)

# Preview blood pressure measurements
display(blood_pressure.head())

# Preview hypertension questionnaire
display(bp_questionnaire.head())
```

```python
# Restrict the study population to adults aged 20 or older
adults = demographics.loc[
    demographics["RIDAGEYR"] >= 20
].copy()

print("Total NHANES participants:", len(demographics))
print("Adult participants (age >= 20):", len(adults))
print("Unique adult participants:", adults["SEQN"].nunique())
```

```python
# Merge demographics with body measurements
nhanes = adults.merge(
    body_measures,
    on="SEQN",
    how="left",
    validate="one_to_one"
)

# Add blood pressure examination data
nhanes = nhanes.merge(
    blood_pressure,
    on="SEQN",
    how="left",
    validate="one_to_one"
)

# Add blood pressure questionnaire data
nhanes = nhanes.merge(
    bp_questionnaire,
    on="SEQN",
    how="left",
    validate="one_to_one"
)

# Verify the merged dataset
print("Merged dataset shape:", nhanes.shape)
print("Unique participants:", nhanes["SEQN"].nunique())
print("Duplicate SEQN:", nhanes["SEQN"].duplicated().sum())
```

```python
# Key variables for our initial analysis
key_variables = [
    "RIDAGEYR",  # Age
    "RIAGENDR",  # Recorded sex
    "BMXBMI",    # Body mass index
    "BMXWAIST",  # Waist circumference
    "BPXSY1",    # First systolic BP reading
    "BPXDI1",    # First diastolic BP reading
    "BPQ020",    # Ever told they had high BP
    "BPQ050A"    # Currently taking prescribed BP medication
]

# Count and calculate percentage of missing values
missing_summary = pd.DataFrame({
    "Missing Count": nhanes[key_variables].isna().sum(),
    "Missing (%)": (
        nhanes[key_variables].isna().mean() * 100
    ).round(2)
})

display(missing_summary)
```

```python
# Identify all available systolic and diastolic BP readings
bp_columns = [
    "BPXSY1", "BPXSY2", "BPXSY3",
    "BPXDI1", "BPXDI2", "BPXDI3"
]

# Check which columns exist
print("Available BP columns:")
print([col for col in bp_columns if col in nhanes.columns])

# Examine missing values across readings
bp_missing = pd.DataFrame({
    "Missing Count": nhanes[bp_columns].isna().sum(),
    "Missing (%)": (
        nhanes[bp_columns].isna().mean() * 100
    ).round(2)
})

display(bp_missing)
```

```python
# Examine responses to the hypertension questionnaire
print("BPQ020: Ever told you had high blood pressure")
print(nhanes["BPQ020"].value_counts(dropna=False))

print("\nBPQ050A: Currently taking prescribed BP medication")
print(nhanes["BPQ050A"].value_counts(dropna=False))

# Cross-tabulate the two variables
print("\nRelationship between BPQ020 and BPQ050A:")

display(
    pd.crosstab(
        nhanes["BPQ020"],
        nhanes["BPQ050A"],
        dropna=False
    )
)
```

```python
# Define blood pressure measurement columns
sbp_cols = ["BPXSY1", "BPXSY2", "BPXSY3"]
dbp_cols = ["BPXDI1", "BPXDI2", "BPXDI3"]

# Inspect zero and negative readings before calculating averages
for col in sbp_cols + dbp_cols:
    print(
        col,
        "Zero:", (nhanes[col] == 0).sum(),
        "Negative:", (nhanes[col] < 0).sum()
    )

# Calculate mean blood pressure using available readings
nhanes["MEAN_SBP"] = nhanes[sbp_cols].mean(axis=1)
nhanes["MEAN_DBP"] = nhanes[dbp_cols].mean(axis=1)

# Count available measurements per participant
nhanes["N_SBP_READINGS"] = nhanes[sbp_cols].notna().sum(axis=1)
nhanes["N_DBP_READINGS"] = nhanes[dbp_cols].notna().sum(axis=1)

# Summarize results
display(
    nhanes[
        [
            "MEAN_SBP",
            "MEAN_DBP",
            "N_SBP_READINGS",
            "N_DBP_READINGS"
        ]
    ].describe()
)
```

```python
# Identify participants with unusually low mean DBP
low_dbp = nhanes.loc[
    nhanes["MEAN_DBP"] < 30,
    [
        "SEQN",
        "RIDAGEYR",
        "BPXDI1",
        "BPXDI2",
        "BPXDI3",
        "MEAN_DBP",
        "MEAN_SBP"
    ]
]

print("Participants with mean DBP < 30:", len(low_dbp))

display(low_dbp)
```

```python
# Count participants by number of available BP readings
reading_summary = (
    nhanes["N_SBP_READINGS"]
    .value_counts()
    .sort_index()
)

display(reading_summary)
```

```python
# Define blood pressure columns
sbp_cols = ["BPXSY1", "BPXSY2", "BPXSY3"]
dbp_cols = ["BPXDI1", "BPXDI2", "BPXDI3"]

bp_cols = sbp_cols + dbp_cols

# Identify extremely small values representing SAS missing data
for col in bp_cols:
    artifact_count = (
        (nhanes[col] > 0) &
        (nhanes[col] < 1e-10)
    ).sum()

    print(f"{col}: {artifact_count} tiny values")

# Replace these values with NaN
nhanes[bp_cols] = nhanes[bp_cols].mask(
    (nhanes[bp_cols] > 0) &
    (nhanes[bp_cols] < 1e-10)
)
```

```python
# Recalculate mean BP using valid available measurements
nhanes["MEAN_SBP"] = nhanes[sbp_cols].mean(axis=1)
nhanes["MEAN_DBP"] = nhanes[dbp_cols].mean(axis=1)

# Recalculate number of available readings
nhanes["N_SBP_READINGS"] = nhanes[sbp_cols].notna().sum(axis=1)
nhanes["N_DBP_READINGS"] = nhanes[dbp_cols].notna().sum(axis=1)

# Examine updated descriptive statistics
display(
    nhanes[
        [
            "MEAN_SBP",
            "MEAN_DBP",
            "N_SBP_READINGS",
            "N_DBP_READINGS"
        ]
    ].describe().round(2)
)
```

```python
print("Systolic readings:")
print(nhanes["N_SBP_READINGS"].value_counts().sort_index())

print("\nDiastolic readings:")
print(nhanes["N_DBP_READINGS"].value_counts().sort_index())

print("\nParticipants with mean DBP < 30:")
print((nhanes["MEAN_DBP"] < 30).sum())
```

```python
# Examine remaining unusually low mean diastolic values
low_dbp = nhanes.loc[
    nhanes["MEAN_DBP"] < 30,
    [
        "SEQN",
        "RIDAGEYR",
        "BPXDI1",
        "BPXDI2",
        "BPXDI3",
        "MEAN_DBP",
        "MEAN_SBP",
        "N_DBP_READINGS"
    ]
]

display(low_dbp)
```

```python
# Create a medication indicator using questionnaire responses

# 1 = currently taking BP medication
# 0 = not currently taking BP medication
# NaN = medication status uncertain

nhanes["BP_MEDICATION"] = np.nan

# Explicitly reported current medication use
nhanes.loc[
    nhanes["BPQ050A"] == 1,
    "BP_MEDICATION"
] = 1

# Explicitly reported no current medication use
nhanes.loc[
    nhanes["BPQ050A"] == 2,
    "BP_MEDICATION"
] = 0

# Participants who reported never being told they had high BP
# were not asked the medication follow-up question.
# Treat as no reported BP medication use for this operational definition.
nhanes.loc[
    nhanes["BPQ020"] == 2,
    "BP_MEDICATION"
] = 0

print(nhanes["BP_MEDICATION"].value_counts(dropna=False))
```

```python
# Determine whether measured BP meets the hypertension threshold
nhanes["ELEVATED_BP"] = (
    (nhanes["MEAN_SBP"] >= 130) |
    (nhanes["MEAN_DBP"] >= 80)
)

# Initialize outcome as missing
nhanes["HYPERTENSION"] = np.nan

# Positive: Elevated measured BP OR current BP medication
nhanes.loc[
    nhanes["ELEVATED_BP"] |
    (nhanes["BP_MEDICATION"] == 1),
    "HYPERTENSION"
] = 1

# Negative: Both BP measurements available and below threshold,
# with no reported current BP medication
negative_mask = (
    nhanes["MEAN_SBP"].notna() &
    nhanes["MEAN_DBP"].notna() &
    (nhanes["MEAN_SBP"] < 130) &
    (nhanes["MEAN_DBP"] < 80) &
    (nhanes["BP_MEDICATION"] == 0)
)

nhanes.loc[negative_mask, "HYPERTENSION"] = 0

# Examine the outcome distribution
print(nhanes["HYPERTENSION"].value_counts(dropna=False))

print("\nHypertension percentages:")
print(
    nhanes["HYPERTENSION"]
    .value_counts(normalize=True, dropna=True) * 100
)
```

```python
print("Total adult participants:", len(nhanes))

print(
    "Participants with known hypertension status:",
    nhanes["HYPERTENSION"].notna().sum()
)

print(
    "Participants with unknown hypertension status:",
    nhanes["HYPERTENSION"].isna().sum()
)

print(
    "Observed hypertension prevalence (%):",
    round(nhanes["HYPERTENSION"].mean() * 100, 2)
)
```

```python
# Verify that all participants with reported BP medication use
# are classified as having hypertension.

medication_positive = nhanes["BP_MEDICATION"] == 1

print(
    "Medication users classified as hypertensive:",
    nhanes.loc[
        medication_positive, "HYPERTENSION"
    ].value_counts(dropna=False)
)

# Verify that all participants meeting measured BP thresholds
# are classified as hypertensive.

elevated_bp = (
    (nhanes["MEAN_SBP"] >= 130) |
    (nhanes["MEAN_DBP"] >= 80)
)

print("\nParticipants with elevated measured BP:")
print(
    nhanes.loc[
        elevated_bp, "HYPERTENSION"
    ].value_counts(dropna=False)
)

# Examine how many participants have missing outcomes
print("\nMissing outcome count:")
print(nhanes["HYPERTENSION"].isna().sum())
```

```python
# Retain participants with a defined hypertension outcome
nhanes_clean = nhanes.loc[
    nhanes["HYPERTENSION"].notna()
].copy()

# Convert target to integer (0 or 1)
nhanes_clean["HYPERTENSION"] = (
    nhanes_clean["HYPERTENSION"].astype(int)
)

print("Final cohort shape:", nhanes_clean.shape)
print("Unique participants:", nhanes_clean["SEQN"].nunique())

print("\nOutcome distribution:")
print(nhanes_clean["HYPERTENSION"].value_counts())
```

```python
# Variables selected for initial exploratory analysis
eda_variables = [
    "RIDAGEYR",  # Age in years
    "BMXBMI",    # Body mass index (kg/m²)
    "BMXWAIST"   # Waist circumference (cm)
]

# Compare descriptive statistics by hypertension status
eda_summary = (
    nhanes_clean
    .groupby("HYPERTENSION")[eda_variables]
    .agg(["count", "mean", "median", "std"])
    .round(2)
)

display(eda_summary)
```

```python
# Create side-by-side boxplots for age, BMI, and waist circumference

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

variables = ["RIDAGEYR", "BMXBMI", "BMXWAIST"]
titles = ["Age", "Body Mass Index", "Waist Circumference"]
ylabels = ["Age (years)", "BMI (kg/m²)", "Waist (cm)"]

for ax, variable, title, ylabel in zip(
    axes, variables, titles, ylabels
):
    sns.boxplot(
        data=nhanes_clean,
        x="HYPERTENSION",
        y=variable,
        ax=ax
    )

    ax.set_title(title)
    ax.set_xlabel("Hypertension Status")
    ax.set_ylabel(ylabel)
    ax.set_xticks([0, 1])
    ax.set_xticklabels(["No Hypertension", "Hypertension"])

plt.suptitle(
    "Demographic and Clinical Characteristics by Hypertension Status",
    fontsize=14
)

plt.tight_layout()
plt.show()
```

```python
# Check missing values for our initial predictor variables
predictor_missing = pd.DataFrame({
    "Missing Count": nhanes_clean[eda_variables].isna().sum(),
    "Missing (%)": (
        nhanes_clean[eda_variables].isna().mean() * 100
    ).round(2)
})

display(predictor_missing)
```

```python
# Calculate Pearson correlations among numerical predictors
correlation_matrix = nhanes_clean[
    ["RIDAGEYR", "BMXBMI", "BMXWAIST"]
].corr()

# Visualize the correlation matrix
plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    square=True
)

plt.title("Correlation Among Numerical Predictors")
plt.tight_layout()
plt.show()
```

```python
# Load smoking and physical activity questionnaires
smoking = pd.read_sas(data_dir / "SMQ_J.XPT")
physical_activity = pd.read_sas(data_dir / "PAQ_J.XPT")

# Inspect dimensions and participant IDs
for name, df in {
    "Smoking": smoking,
    "Physical Activity": physical_activity
}.items():
    print(f"\n{name}")
    print("Shape:", df.shape)
    print("Unique participants:", df["SEQN"].nunique())
    print("Duplicate SEQN:", df["SEQN"].duplicated().sum())
    print("Columns:", df.columns.tolist())
```

```python
# Examine smoking questionnaire responses
smoking_vars = ["SMQ020", "SMQ040"]

for col in smoking_vars:
    print(f"\n{col}")
    print(smoking[col].value_counts(dropna=False))

# Examine physical activity questionnaire responses
activity_vars = [
    "PAQ605",
    "PAQ620",
    "PAQ650",
    "PAQ665"
]

for col in activity_vars:
    print(f"\n{col}")
    print(physical_activity[col].value_counts(dropna=False))
```

```python
# Create smoking status from lifetime and current smoking responses

smoking["SMOKING_STATUS"] = pd.NA

# Never smoker
smoking.loc[
    smoking["SMQ020"] == 2,
    "SMOKING_STATUS"
] = "Never"

# Former smoker
smoking.loc[
    (smoking["SMQ020"] == 1) &
    (smoking["SMQ040"] == 3),
    "SMOKING_STATUS"
] = "Former"

# Current smoker: every day or some days
smoking.loc[
    (smoking["SMQ020"] == 1) &
    (smoking["SMQ040"].isin([1, 2])),
    "SMOKING_STATUS"
] = "Current"

print(smoking["SMOKING_STATUS"].value_counts(dropna=False))
```

```python
# Identify four types of reported physical activity
activity_cols = [
    "PAQ605",
    "PAQ620",
    "PAQ650",
    "PAQ665"
]

# Initialize as missing
physical_activity["ANY_PHYSICAL_ACTIVITY"] = np.nan

# At least one Yes response
any_yes = physical_activity[activity_cols].eq(1).any(axis=1)

# All four responses explicitly No
all_no = physical_activity[activity_cols].eq(2).all(axis=1)

physical_activity.loc[
    any_yes,
    "ANY_PHYSICAL_ACTIVITY"
] = 1

physical_activity.loc[
    all_no,
    "ANY_PHYSICAL_ACTIVITY"
] = 0

print(
    physical_activity["ANY_PHYSICAL_ACTIVITY"]
    .value_counts(dropna=False)
)
```

```python
# Merge smoking status
nhanes_clean = nhanes_clean.merge(
    smoking[["SEQN", "SMOKING_STATUS"]],
    on="SEQN",
    how="left",
    validate="one_to_one"
)

# Merge physical activity indicator
nhanes_clean = nhanes_clean.merge(
    physical_activity[["SEQN", "ANY_PHYSICAL_ACTIVITY"]],
    on="SEQN",
    how="left",
    validate="one_to_one"
)

# Verify the cohort remains unchanged
print("Final cohort shape:", nhanes_clean.shape)
print("Unique participants:", nhanes_clean["SEQN"].nunique())
print("Duplicate participants:", nhanes_clean["SEQN"].duplicated().sum())
```

```python
print("Smoking status:")
print(
    nhanes_clean["SMOKING_STATUS"]
    .value_counts(dropna=False)
)

print("\nPhysical activity:")
print(
    nhanes_clean["ANY_PHYSICAL_ACTIVITY"]
    .value_counts(dropna=False)
)

# Missingness summary for our current predictor set
predictors = [
    "RIDAGEYR",
    "RIAGENDR",
    "BMXBMI",
    "BMXWAIST",
    "SMOKING_STATUS",
    "ANY_PHYSICAL_ACTIVITY"
]

missing_summary = pd.DataFrame({
    "Missing Count": nhanes_clean[predictors].isna().sum(),
    "Missing (%)": (
        nhanes_clean[predictors].isna().mean() * 100
    ).round(2)
})

display(missing_summary)
```

```python
# Calculate hypertension proportion within each smoking category
smoking_summary = (
    nhanes_clean
    .groupby("SMOKING_STATUS")["HYPERTENSION"]
    .agg(["count", "sum", "mean"])
    .rename(columns={
        "count": "Total",
        "sum": "Hypertension Cases",
        "mean": "Hypertension Rate"
    })
)

smoking_summary["Hypertension Rate"] *= 100

print("Hypertension by smoking status:")
display(smoking_summary.round(2))


# Calculate hypertension proportion by physical activity
activity_summary = (
    nhanes_clean
    .groupby("ANY_PHYSICAL_ACTIVITY")["HYPERTENSION"]
    .agg(["count", "sum", "mean"])
    .rename(columns={
        "count": "Total",
        "sum": "Hypertension Cases",
        "mean": "Hypertension Rate"
    })
)

activity_summary["Hypertension Rate"] *= 100

print("Hypertension by physical activity:")
display(activity_summary.round(2))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Smoking status
sns.barplot(
    data=smoking_summary.reset_index(),
    x="SMOKING_STATUS",
    y="Hypertension Rate",
    order=["Never", "Former", "Current"],
    errorbar=None,
    ax=axes[0]
)

axes[0].set_title("Hypertension by Smoking Status")
axes[0].set_xlabel("Smoking Status")
axes[0].set_ylabel("Hypertension (%)")
axes[0].set_ylim(0, 100)

# Physical activity
sns.barplot(
    data=activity_summary.reset_index(),
    x="ANY_PHYSICAL_ACTIVITY",
    y="Hypertension Rate",
    errorbar=None,
    ax=axes[1]
)

axes[1].set_title("Hypertension by Reported Physical Activity")
axes[1].set_xlabel("Any Selected Physical Activity")
axes[1].set_ylabel("Hypertension (%)")
axes[1].set_xticks([0, 1])
axes[1].set_xticklabels(["No", "Yes"])
axes[1].set_ylim(0, 100)

plt.suptitle(
    "Hypertension Classification by Lifestyle Characteristics",
    fontsize=14
)

plt.tight_layout()
plt.show()
```

```python
# Compare age distributions across lifestyle categories

smoking_age = (
    nhanes_clean
    .groupby("SMOKING_STATUS")["RIDAGEYR"]
    .agg(["count", "mean", "median"])
    .round(2)
)

activity_age = (
    nhanes_clean
    .groupby("ANY_PHYSICAL_ACTIVITY")["RIDAGEYR"]
    .agg(["count", "mean", "median"])
    .round(2)
)

print("Age by smoking status:")
display(smoking_age)

print("\nAge by physical activity:")
display(activity_age)
```

```python
# Define our final predictor variables
feature_cols = [
    "RIDAGEYR",
    "RIAGENDR",
    "BMXBMI",
    "BMXWAIST",
    "SMOKING_STATUS",
    "ANY_PHYSICAL_ACTIVITY"
]

# Target variable
target_col = "HYPERTENSION"

# Include SEQN for tracking participants, not as a predictor
final_cols = ["SEQN"] + feature_cols + [target_col]

# Create an independent modeling dataset
model_data = nhanes_clean[final_cols].copy()

print("Modeling dataset shape:", model_data.shape)

display(model_data.head())
```

```python
# Check categorical variable values
categorical_cols = [
    "RIAGENDR",
    "SMOKING_STATUS",
    "ANY_PHYSICAL_ACTIVITY"
]

for col in categorical_cols:
    print(f"\n{col}:")
    print(model_data[col].value_counts(dropna=False))
```

```python
from pathlib import Path

# Create output directory if necessary
output_dir = Path("Outputs")
output_dir.mkdir(parents=True, exist_ok=True)

# Save the modeling dataset
output_path = output_dir / "NHANES_Hypertension_Model_Data.csv"

model_data.to_csv(output_path, index=False)

print("Dataset saved to:", output_path)
```
